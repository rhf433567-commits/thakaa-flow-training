## التوقع قبل التشغيل

قبل تشغيل الكود، سأقرأ كل سطر وأكتب توقعي للنتيجة، ثم أقارن التوقع بالنتيجة الفعلية وأوضح سبب أي اختلاف.

 جدول التوقعات والنتائج الفعلية

| **السطر البرمجي** | **توقعي قبل التشغيل** | **الناتج الفعلي** | **سبب الاختلاف** |
|---|---|---|---|
| `type(df["score"])` | Series | `pandas.core.series.Series` | لا يوجد اختلاف |
| `type(df[["score"]])` | DataFrame | `pandas.core.frame.DataFrame` | لا يوجد اختلاف |
| `df.loc["b", "city"]` | Jeddah | Jeddah | لا يوجد اختلاف |
| `df.iloc[1, 1]` | Jeddah | Jeddah | لا يوجد اختلاف |
| `df.loc["a":"c", "score"].tolist()` | [88, 92, 75] | [88, 92, 75] | لا يوجد اختلاف |
| `df.iloc[0:2, 2].tolist()` | [88, 92] | [88, 92] | لا يوجد اختلاف |
| `df[df["score"] > 80]["name"].tolist()` | ['Sara', 'Noura'] | ['Sara', 'Noura'] | لا يوجد اختلاف |
| `df["city"].value_counts()["Riyadh"]` | 2 | 2 | لا يوجد اختلاف |

### ملاحظات على التوقعات

- القوس الواحد `df["score"]` يعيد Series، أي عمودًا واحدًا على شكل سلسلة.
- القوسان `df[["score"]]` يعيدان DataFrame، أي جدولًا يحتفظ بشكل الجدول حتى لو كان يحتوي على عمود واحد.
- `loc` يعتمد على أسماء الصفوف والأعمدة.
- `iloc` يعتمد على مواقع الصفوف والأعمدة بالأرقام.
- `loc` يشمل نهاية النطاق عند تحديد أسماء الصفوف، بينما `iloc` لا يشمل نهاية النطاق.
- الفلترة تساعدنا على استخراج البيانات التي تحقق شرطًا معينًا.
- `value_counts()` يحسب عدد مرات تكرار كل قيمة.

### رأيي في النتائج

```python
print(type(df["score"]))
print(type(df[["score"]]))
print(df.loc["b", "city"])
print(df.iloc[1, 1])
print(df.loc["a":"c", "score"].tolist())
print(df.iloc[0:2, 2].tolist())
print(df[df["score"] > 80]["name"].tolist())
print(df["city"].value_counts()["Riyadh"])
```

 1. الفرق بين Series وDataFrame

تعلمت أن استخدام قوس واحد مثل `df["score"]` يعطيني عمودًا واحدًا من نوع Series، أما استخدام قوسين مثل `df[["score"]]` فيعطيني DataFrame يحتفظ بشكل الجدول.

أستخدم Series عندما أحتاج التعامل مع عمود واحد، وأستخدم DataFrame عندما أحتاج الاحتفاظ بشكل الجدول أو اختيار أكثر من عمود.

 2. لماذا loc يعيد 3 صفوف بينما iloc يعيد صفين؟

السبب هو اختلاف طريقة تحديد النطاق.

`loc` يعتمد على أسماء الصفوف ويشمل آخر اسم في النطاق، لذلك أعاد الصفوف a وb وc وعددها 3.

أما `iloc` فيعتمد على أرقام المواقع ولا يشمل موقع النهاية، لذلك أعاد الصفين الأول والثاني فقط وعددهما 2.

تعلمت أن الفرق بينهما مهم عند اختيار البيانات حتى لا أحصل على صفوف أكثر أو أقل من المطلوب.

 3. الفرق بين Pandas وNumPy

Pandas تُستخدم لتنظيم البيانات في جداول تحتوي على صفوف وأعمدة، وتساعدني في قراءة البيانات وتصفيتها وتحليلها.

أما NumPy فتُستخدم في العمليات الحسابية والتعامل مع المصفوفات والأرقام.

نحتاج NumPy رغم وجود Pandas لأن NumPy متخصصة في العمليات الرقمية، وكثير من وظائف Pandas تعتمد عليها. لذلك أرى أن المكتبتين تكملان بعضهما في تحليل البيانات.

 4. تطبيق Pandas على الحياة اليومية

اخترت مثال جدول المصروفات الشهرية.

- الصف يمثل عملية مصروف واحدة، مثل شراء الطعام أو دفع المواصلات.
- العمود يمثل نوعًا من المعلومات، مثل نوع المصروف أو المبلغ أو التاريخ.
- العمود الذي قد يكون ناقصًا هو طريقة الدفع، مثل النقد أو البطاقة.

أرى أن إضافة طريقة الدفع تساعدني على تنظيم مصروفاتي ومعرفة كيف أنفق أموالي بشكل أفضل.

ومن خلال هذا المثال فهمت أن Pandas تساعدني في تنظيم وتحليل البيانات حتى لو كانت بيانات بسيطة من حياتي اليومية.

مثال عملي: جدول المصروفات الشهرية

| نوع المصروف | المبلغ | التاريخ |
|---|---:|---|

| التسوق | 200 | 5 نوفمبر |

- الصف يمثل عملية مصروف واحدة.
- العمود يمثل نوعًا من المعلومات، مثل المبلغ أو التاريخ.
- العمود الذي قد يكون ناقصًا هو طريقة الدفع؛ لأنها توضح هل تم الدفع نقدًا أو بالبطاقة.

# تنظيف البيانات ومعالجة القيم المفقودةTask 3

```python
clean_df = df.copy()

print("Original data copied successfully!")
display(clean_df)
```

```python
print("القيم المفقودة في كل عمود:")
print(clean_df.isnull().sum())

print("\nعدد الصفوف المكررة:")
print(clean_df.duplicated().sum())
```

```python
print("أرقام الطلبات المكررة:")
print(clean_df.loc[clean_df.duplicated(), "order_id"].tolist())

clean_df = clean_df.drop_duplicates()

print("عدد الصفوف بعد حذف التكرار:", len(clean_df))
print("عدد الصفوف المكررة المتبقية:", clean_df.duplicated().sum())
```

```python
clean_df["city"] = clean_df["city"].str.strip().str.title()

print("عدد الطلبات في كل مدينة:")
print(clean_df["city"].value_counts(dropna=False))
```

 توحيد أسماء المدن

لاحظت أن أسماء المدن مكتوبة بأكثر من طريقة، مثل `Riyadh` و`riyadh` و`RIYADH`، وهذا قد يؤدي إلى حساب المدينة نفسها كأنها مدن مختلفة.

استخدمت الدالة `strip()` لإزالة المسافات الزائدة، والدالة `title()` لتوحيد طريقة كتابة الأحرف.

بعد تنفيذ الخطوة، أصبحت أسماء المدن موحدة، وظهر أن مدينة واحدة ما زالت قيمتها مفقودة، وسأعالجها في خطوات لاحقة.

```python
clean_df["price"] = clean_df["price"].replace("eighteen", "18")
clean_df["price"] = clean_df["price"].astype(float)

print("نوع بيانات السعر:")
print(clean_df["price"].dtype)

print("الأسعار:")
print(clean_df["price"].unique())
```


لاحظت وجود قيمة غير رقمية في عمود السعر، وهي `eighteen` بدلًا من الرقم `18`.

قمت بتصحيح هذه القيمة باستخدام `replace()`، ثم استخدمت `astype(float)` لتحويل جميع الأسعار إلى أرقام عشرية.

الهدف من هذه الخطوة هو التأكد من أن عمود السعر يحتوي على بيانات رقمية صحيحة، حتى يمكن استخدامه في العمليات الحسابية والتحليل دون أخطاء.

بعد تنفيذ الخطوة، أصبح نوع البيانات `float64`، وهذا يدل على نجاح عملية التحويل.

```python
invalid_ratings = clean_df["rating"].notna() & ~clean_df["rating"].between(1, 5)

print("التقييمات غير الصحيحة:")
display(clean_df.loc[invalid_ratings, ["order_id", "rating"]])

clean_df.loc[invalid_ratings, "rating"] = None

print("عدد التقييمات غير الصحيحة المتبقية:")
print((clean_df["rating"].notna() & ~clean_df["rating"].between(1, 5)).sum())
```

 اكتشاف التقييمات غير الصحيحة ومعالجتها

قمت بفحص عمود التقييمات للتأكد من أن جميع القيم تقع بين 1 و5، ولاحظت وجود تقييم غير صحيح قيمته 30 في الطلب رقم 110.

اعتبرت هذه القيمة غير صالحة لأنها خارج النطاق المحدد للتقييمات، لذلك قمت بتحويلها إلى قيمة مفقودة حتى أتمكن من تعويضها بطريقة مناسبة في الخطوات القادمة.

بعد تنفيذ العملية، لم تعد توجد أي تقييمات غير صحيحة خارج النطاق.

```python
quantity_median = clean_df["quantity"].median()

print("وسيط الكمية:")
print(quantity_median)
```

 حساب وسيط الكمية واختيار طريقة التعويض

قمت بحساب وسيط الكمية، وكانت النتيجة 1.5.

بعد ذلك قررت استخدام المنوال بدلًا من الوسيط لتعويض القيمة المفقودة؛ لأن الكمية تمثل عددًا صحيحًا من المشروبات، ولا يمكن أن تكون 1.5 مشروب.

لذلك اخترت المنوال لأنه يمثل الكمية الأكثر تكرارًا في البيانات، ويساعد على تعويض القيمة الناقصة بقيمة منطقية.

```python
quantity_mode = clean_df["quantity"].mode()[0]

clean_df["quantity"] = clean_df["quantity"].fillna(quantity_mode)

print("المنوال المستخدم:", quantity_mode)
print("القيم المفقودة في الكمية:", clean_df["quantity"].isna().sum())
```

 تعويض الكمية المفقودة

بعد حساب المنوال، وجدت أن أكثر كمية تكرارًا هي 1، لذلك استخدمتها لتعويض القيمة المفقودة في عمود الكمية.

اخترت المنوال بدلًا من الوسيط لأن الكمية تمثل عدد المشروبات، ويجب أن تكون عددًا صحيحًا.

بعد التعويض، أصبح عمود الكمية خاليًا من القيم المفقودة.

```python
rating_median = clean_df["rating"].median()
clean_df["rating"] = clean_df["rating"].fillna(rating_median)

drink_mode = clean_df["drink"].mode()[0]
clean_df["drink"] = clean_df["drink"].fillna(drink_mode)

clean_df["city"] = clean_df["city"].fillna("Unknown")

print("وسيط التقييم المستخدم:", rating_median)
print("منوال المشروب المستخدم:", drink_mode)

print("\nالقيم المفقودة المتبقية:")
print(clean_df.isna().sum())
```

### معالجة القيم المفقودة في التقييم والمدينة والمشروب

قمت بمعالجة القيم المفقودة في ثلاثة أعمدة باستخدام طرق مختلفة حسب طبيعة البيانات.

في عمود التقييم استخدمت الوسيط وكانت قيمته 4، لأنه مناسب للبيانات الرقمية ويساعد على تقليل تأثير القيم المتطرفة.

في عمود المشروب استخدمت المنوال، وكانت قيمته Latte لأنه المشروب الأكثر تكرارًا في البيانات.

أما في عمود المدينة فاستخدمت القيمة Unknown بدلًا من اختيار مدينة بشكل عشوائي؛ لأن المدينة الأصلية غير معروفة.

بعد تنفيذ هذه الخطوات، أصبحت جميع الأعمدة خالية من القيم المفقودة.

```python
print("القيم المفقودة في كل عمود:")
print(clean_df.isna().sum())

print("\nإجمالي القيم المفقودة:")
print(clean_df.isna().sum().sum())

print("\nالبيانات بعد التنظيف:")
display(clean_df)
```

```python
trips = {
    "traveler": ["Zara", "Omar", "Lina", "Fahad", "Zara", "Nora", "Yousef", "Salma", "Khalid", "Maya"],
    "departure_year": [2026, 2026, 2026, 2026, 2026, 2026, 2026, 2026, 2026, 2026],
    "destination_year": ["1450", "3026", "-300", "1999", "1450", "2500", "1000000", "1917", "1450", None],
    "age": [29, 35, 41, 27, 29, 33, 38, -5, 45, 31],
    "price": [12000, 30000, 45000, 8000, 12000, 22000, 999999, 15000, 12000, 9000],
    "status": ["done", "done", "cancelled", "done", "done", "DONE", "done", "done", "Done", "done"],
}

trips_df = pd.DataFrame(trips)

display(trips_df)
```

ومن خلال فحص البيانات، لاحظت وجود عمر غير منطقي وهو -5، بالإضافة إلى سنة وجهة غير معتادة وهي 1000000.

أرى أن هذه المشكلات تحتاج إلى معالجة قبل البدء في تحليل الرحلات حتى تكون النتائج أكثر دقة.

```python
print("عدد الصفوف المكررة:")
print(trips_df.duplicated().sum())

print("\nالصفوف المكررة:")
display(trips_df[trips_df.duplicated()])
```

 فحص البيانات المكررة

وجدت صفًا مكررًا واحدًا للعميلة Zara، حيث ظهرت الرحلة بنفس سنة المغادرة والوجهة والعمر والسعر والحالة.

قررت حذف الصف المكرر حتى لا يتم احتساب الرحلة مرتين، مما يساعد على دقة التحليل.

```python
trips_clean = trips_df.copy()

trips_clean = trips_clean.drop_duplicates()

trips_clean["status"] = trips_clean["status"].str.lower()

print("عدد الصفوف بعد حذف التكرار:")
print(len(trips_clean))

print("\nحالات الرحلات بعد توحيد الكتابة:")
print(trips_clean["status"].value_counts())
```

 حذف التكرار وتوحيد حالات الرحلات

حذفت الصف المكرر باستخدام drop_duplicates، فأصبح عدد الرحلات 9 بدلًا من 10.

كما وحدت كتابة حالات الرحلات باستخدام lower()، حتى تظهر جميع الحالات بنفس الشكل. أصبحت لدينا 8 رحلات مكتملة ورحلة واحدة ملغاة.

هذا يساعد على منع تكرار البيانات وتسهيل تحليل الحالات.

```python
trips_clean["destination_year"] = pd.to_numeric(
    trips_clean["destination_year"], errors="coerce"
)

trips_clean.loc[trips_clean["age"] < 0, "age"] = None

print("القيم المفقودة بعد التحويل:")
print(trips_clean.isnull().sum())

print("\nالبيانات بعد التحويل:")
display(trips_clean)
```

```python
trips_clean = trips_clean.dropna(subset=["destination_year"])

median_age = trips_clean["age"].median()

trips_clean["age"] = trips_clean["age"].fillna(median_age)

print("وسيط الأعمار:", median_age)

print("\nالقيم المفقودة بعد المعالجة:")
print(trips_clean.isnull().sum())

display(trips_clean)
```

 معالجة القيم المفقودة والأعمار غير الصحيحة

قررت حذف رحلة Maya بسبب عدم وجود سنة الوجهة، لأن هذه المعلومة ضرورية لتحليل الرحلة.

أما عمر Salma فكان -5، وهو عمر غير منطقي، لذلك اعتبرته قيمة مفقودة وعوضته بوسيط الأعمار، وكان الوسيط 35 سنة.

بعد المعالجة أصبحت جميع الأعمدة خالية من القيم المفقودة.

 التعامل مع سنة الوجهة غير المعتادة

لاحظت أن رحلة Yousef وجهتها سنة 1000000، وسعرها 999999.

قررت الاحتفاظ بهذه الرحلة بدل حذفها؛ لأن فكرة المشروع تعتمد على السفر عبر الزمن، وقد تكون الرحلة إلى المستقبل البعيد مقصودة، كما أن ارتفاع السعر قد يكون مرتبطًا ببعد الوجهة.

```python
trips_clean["years_traveled"] = abs(
    trips_clean["destination_year"] - trips_clean["departure_year"]
)

trips_clean["direction"] = trips_clean["destination_year"].apply(
    lambda year: "past" if year < 2026 else "future"
)

display(trips_clean)
```

 إنشاء أعمدة مدة الرحلة واتجاهها

أنشأت عمودًا لحساب عدد السنوات بين سنة المغادرة وسنة الوجهة، وعمودًا آخر لتحديد اتجاه الرحلة.

استخدمت past للرحلات إلى الماضي وfuture للرحلات إلى المستقبل.

ساعدتني هذه الأعمدة على معرفة مدة كل رحلة وتصنيف الرحلات حسب اتجاهها، وأظهرت النتائج أن رحلة Yousef هي الأطول.

```python
print("أكثر سنة وجهة طلبًا:")
print(trips_clean["destination_year"].value_counts().idxmax())

print("\nمتوسط سعر الرحلات حسب الاتجاه:")
print(trips_clean.groupby("direction")["price"].mean())

print("\nعدد الرحلات المكتملة:")
print((trips_clean["status"] == "done").sum())

print("\nإيرادات الرحلات المكتملة:")
print(trips_clean.loc[trips_clean["status"] == "done", "price"].sum())

print("\nصاحب أطول رحلة:")
longest_trip = trips_clean.loc[trips_clean["years_traveled"].idxmax()]
print(longest_trip["traveler"])
print("مدة الرحلة:", longest_trip["years_traveled"])
```
