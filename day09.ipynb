 Task 1  المتوسط يكذب أحيانًا

```python
salaries = np.array([6000, 6500, 7000, 7200, 7500, 8000, 95000])

mean = np.mean(salaries)
median = np.median(salaries)

print("Mean:", mean)
print("Median:", median)
```

 تحليل النتائج

لاحظت أن المتوسط الحسابي 19600 ريال، بينما الوسيط 7200 ريال.

السبب هو وجود راتب المدير التنفيذي المرتفع، والذي أثر على المتوسط ورفعه بشكل كبير.

استنتجت أن الوسيط يمثل رواتب الموظفين بشكل أقرب من المتوسط في هذه الحالة.

```python
Q1 = np.percentile(salaries, 25)
Q3 = np.percentile(salaries, 75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower:", lower)
print("Upper:", upper)

outliers = salaries[(salaries < lower) | (salaries > upper)]

print("Outliers:", outliers)
```

 تحليل القيم الشاذة

حسبت الربع الأول والربع الثالث، ثم استخدمت IQR لتحديد حدود القيم الطبيعية.

ظهرت حدود القيم من 5250 إلى 9250 ريال.

وجدت أن راتب 95000 ريال خارج هذه الحدود، لذلك يعتبر قيمة شاذة (Outlier)، وهو راتب المدير التنفيذي.

```python
salaries_without_ceo = salaries[:-1]

print("Mean without CEO:", np.mean(salaries_without_ceo))
print("Median without CEO:", np.median(salaries_without_ceo))
```

بعد حذف راتب المدير التنفيذي انخفض المتوسط من 19600 إلى 7033.33 ريال بينما تغير الوسيط من 7200 إلى 7100 فقط.

 لماذا نستخدم Scaling؟

نستخدم Scaling لأن الأعمدة تختلف في قيمها ومقاييسها.
مثلاً العمر أرقامه صغيرة مقارنة بالتكاليف.
لذلك نوحّد المقاييس حتى لا يؤثر العمود ذو الأرقام الكبيرة بشكل أكبر على بعض نماذج التعلم الآلي.

```python
charges_range = np.max(charges) - np.min(charges)
age_range = np.max(age) - np.min(age)

print("Charges range:", charges_range)
print("Age range:", age_range)
print("Ratio:", charges_range / age_range)
```

 Task 5 – صيد الكواكب

```python
rng = np.random.default_rng(7)

brightness = 1000 + rng.normal(0, 1, 240)

for start in [30, 102, 174]:
    brightness[start:start + 5] -= 20
```

```python
print("Mean brightness:", np.mean(brightness))
print("Median brightness:", np.median(brightness))
```

```python
brightness_z = (brightness - np.mean(brightness)) / np.std(brightness)

mask = brightness_z < -3

hours = np.arange(240)

print("Number of low readings:", np.sum(mask))
print("Hours:", hours[mask])
```

```python
starts = np.where(np.diff(mask.astype(int)) == 1)[0] + 1

print("Transit start hours:", starts)
```

```python
time_between = np.diff(starts)

print("Time between transits (hours):", time_between)
print("Time between transits (days):", time_between / 24)
```

```python
normal_brightness = brightness[~mask]
low_brightness = brightness[mask]

drop = np.mean(normal_brightness) - np.mean(low_brightness)

percentage = (drop / np.mean(normal_brightness)) * 100

print("Brightness drop:", drop)
print("Drop percentage:", percentage, "%")
```
