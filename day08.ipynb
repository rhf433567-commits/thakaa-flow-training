# Task 3  تحليل مبيعات المقهى

```python
days = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]

sales = np.array([1240, 980, 1105, 1320, 1875, 2410, 2150])

print("Total:", np.sum(sales))
print("Mean:", np.mean(sales))
print("Median:", np.median(sales))
print("Maximum:", np.max(sales))
print("Minimum:", np.min(sales))
print("Standard Deviation:", np.std(sales))
```

**النتائج:**

- إجمالي المبيعات: 11080
- المتوسط: 1582.86
- الوسيط: 1320
- أعلى مبيعات: 2410
- أقل مبيعات: 980
- الانحراف المعياري: 516.80

**ملاحظتي:** يوم الجمعة كان أعلى يوم في المبيعات، ويوم الاثنين كان الأقل.

```python
best_day = np.argmax(sales)
worst_day = np.argmin(sales)

print("Best day:", days[best_day], sales[best_day])
print("Worst day:", days[worst_day], sales[worst_day])
```

استخدمت Boolean Mask لمعرفة الأيام اللي مبيعاتها أعلى من المتوسط، وظهرت عندي 3 أيام وهي الخميس والجمعة والسبت.

```python
above_average = sales > np.mean(sales)

print("Mask:", above_average)
print("Number of days:", np.sum(above_average))
print("Sales:", sales[above_average])
```

استخدمت Boolean Mask لمقارنة المبيعات بالمتوسط، وطلعت عندي 3 أيام مبيعاتها أعلى من المتوسط، وهي الخميس والجمعة والسبت.

```python
weekend_sales = sales[5:7]

print("Weekend sales:", weekend_sales)
print("Weekend total:", np.sum(weekend_sales))
print("Weekend percentage:", np.sum(weekend_sales) / np.sum(sales) * 100)
```

```python
sales_after_tax = sales * 1.15
next_week_target = sales * 1.10

print("Sales after 15% tax:", sales_after_tax)
print("Next week target:", next_week_target)
```

حسبت المبيعات بعد إضافة 15%، وحسبت هدف الأسبوع القادم بزيادة 10% عن المبيعات الحالية، واستخدمت NumPy لتنفيذ العمليات على جميع الأيام مرة واحدة.

```python
changes = np.diff(sales)

print("Daily changes:", changes)
print("Biggest increase:", np.max(changes))
print("Day:", days[np.argmax(changes) + 1])
```

استخدمت `np.diff` لمعرفة التغير في المبيعات بين الأيام، وكانت أكبر زيادة يوم الخميس بمقدار 555 مقارنة بيوم الأربعاء.

أشوف أن الوسيط يمثل المبيعات اليومية المعتادة بشكل أفضل؛ لأن بعض الأيام كانت مبيعاتها مرتفعة، وهذا رفع المتوسط، بينما الوسيط أقل تأثرًا بالقيم المرتفعة.

# Task 4 – تحليل مبيعات الفروع

```python
branches = ["Riyadh", "Jeddah", "Dammam"]

days = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]

sales = np.array([
    [1240, 980, 1105, 1320, 1875, 2410, 2150],
    [910, 870, 990, 1010, 1430, 1980, 2240],
    [640, 720, 680, 750, 990, 1210, 1080]
])

print("Shape:", sales.shape)
```

المصفوفة فيها 3 صفوف تمثل الفروع، و7 أعمدة تمثل أيام الأسبوع، يعني عندنا مبيعات 3 فروع خلال 7 أيام.

```python
print("Jeddah Thursday:", sales[1, 4])

print("Dammam sales:", sales[2, :])

print("Friday sales:", sales[:, 5])

print("Friday and Saturday:")
print(sales[:, 5:7])
```

استخدمت الفهرسة للوصول إلى مبيعات يوم وفرع محدد، واستخدمت Slicing لاستخراج مبيعات أيام أو فروع كاملة.

```python
branch_totals = np.sum(sales, axis=1)
day_totals = np.sum(sales, axis=0)

print("Branch totals:", branch_totals)
print("Day totals:", day_totals)
```

استخدمت `axis=1` لحساب مجموع مبيعات كل فرع، واستخدمت `axis=0` لحساب مجموع مبيعات كل يوم لجميع الفروع.

```python
best_branch = np.argmax(branch_totals)
best_company_day = np.argmax(day_totals)

print("Best branch:", branches[best_branch])
print("Best day:", days[best_company_day])

print("Branch sales percentages:")
print(branch_totals / np.sum(branch_totals) * 100)
```

فرع الرياض حقق أعلى مبيعات بنسبة 41.69%، وكان يوم الجمعة أعلى يوم في المبيعات لجميع الفروع. أما فرع الدمام فكانت مبيعاته الأقل.

```python
max_index = np.argmax(sales)

row = max_index // sales.shape[1]
col = max_index % sales.shape[1]

print("Highest sales:", sales[row, col])
print("Branch:", branches[row])
print("Day:", days[col])
```

استخدمت `argmax` لمعرفة موقع أعلى قيمة في المصفوفة، وبعدها استخدمت القسمة الصحيحة والباقي لتحديد الفرع واليوم. وكانت أعلى مبيعات في الرياض يوم الجمعة بقيمة 2410.

أقترح إغلاق فرع الدمام يوم الأحد لأن مبيعاته كانت الأقل بين الفروع الثلاثة، لكن الأفضل مراجعة عوامل أخرى قبل اتخاذ القرار النهائي.

# Task 5  تحليل بيانات الغواصة

```python
import numpy as np

depth = np.array([
    120, 340, 610, -1, 890, 1150, 1420, 99999, 1730, 2010,
    2280, -1, 2560, 2840, 99999, 3120, 3390, 3650, -1, 3900
])

bad_mask = (depth == -1) | (depth == 99999)

print("Bad readings:", depth[bad_mask])
print("Number of bad readings:", np.sum(bad_mask))
print("Percentage:", np.mean(bad_mask) * 100)
```

```python
valid_mask = ~bad_mask

mean_with_errors = np.mean(depth)
mean_without_errors = np.mean(depth[valid_mask])

print("Mean with errors:", mean_with_errors)
print("Mean without errors:", mean_without_errors)
print("How many times:", mean_with_errors / mean_without_errors)
```

حساب الوسيط وأكبر عمق:

بعد تنظيف البيانات واستبعاد القراءات الخاطئة، حسبت الوسيط وأكبر عمق وصلت له الغواصة بالمتر، وبعدها حولت القيم إلى القدم باستخدام معامل التحويل 3.28084؛ عشان أعرف العمق بوحدتين مختلفتين.

```python
valid_depth = depth[valid_mask]

median_m = np.median(valid_depth)
max_m = np.max(valid_depth)

print("Median in meters:", median_m)
print("Maximum in meters:", max_m)

print("Median in feet:", median_m * 3.28084)
print("Maximum in feet:", max_m * 3.28084)
```

حساب ضغط الماء:

حسبت ضغط الماء عند كل قراءة صحيحة باستخدام المعادلة (العمق ÷ 10) + 1، وبعدها حسبت عدد القراءات اللي تجاوز فيها الضغط 300 ضغط جوي.

```python
pressure = valid_depth / 10 + 1

print("Pressure:", pressure)
print("Readings above 300 atm:", np.sum(pressure > 300))
```

تحليل التغير في العمق:

استخدمت np.diff لحساب الفرق بين كل قراءة صحيحة والقراءة التي بعدها، ولاحظت أن مقدار التغير متقارب بشكل عام، لذلك يبدو أن الغواصة تنزل بمعدل شبه ثابت. لكن لا يمكن تحديد السرعة بالزمن بدون معرفة الفاصل الزمني بين القراءات.

```python
changes = np.diff(valid_depth)

print("Depth changes:", changes)
print("Average change:", np.mean(changes))
```

القرار

سأحوّل القراءات الخاطئة مثل -1 و99999 إلى قيم مفقودة (NaN) بدل ما أستخدمها في التحليل. وإذا احتجت أعوّضها، أستخدم الاستيفاء بالاعتماد على القراءات القريبة والوقت بينها؛ لأن استخدام المتوسط العام ممكن يعطي قيمة غير واقعية لعمق الغواصة.
