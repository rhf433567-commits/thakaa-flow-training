 1. الفرق بين Feature Engineering و Feature Selection

Feature Engineering هو إنشاء أعمدة جديدة من البيانات الموجودة حتى تساعد الموديل على فهم البيانات بشكل أفضل.

Feature Selection هو اختيار الأعمدة المهمة للموديل وحذف الأعمدة التي لا تضيف فائدة.

مثال من مجال الأمن السيبراني

في مشروع اكتشاف روابط التصيد الإلكتروني:

- Target: تحديد هل الرابط تصيدي أو آمن.
- Feature جديدة: حساب طول الرابط (URL Length) من الرابط الأصلي.
- Feature يمكن حذفها: رقم تعريف الرابط (ID) لأنه مجرد معرّف ولا يساعد في التنبؤ.

 2. عائلات Feature Selection

| الطريقة | الفكرة | مثال | الميزة | العيب |
|---|---|---|---|---|
| Filter | اختيار الأعمدة حسب علاقتها بالهدف باستخدام الإحصائيات | Correlation | سريعة وسهلة | قد لا تكتشف العلاقات المعقدة |
| Wrapper | تجربة مجموعات مختلفة من الأعمدة باستخدام الموديل | RFE | تختار الأعمدة بناءً على أداء الموديل | بطيئة وتحتاج وقتًا |
| Embedded | اختيار الأعمدة أثناء تدريب الموديل | Lasso | تجمع بين التدريب واختيار الأعمدة | تعتمد على نوع الموديل |

نستخدم Filter هذا الأسبوع لأنها سهلة وسريعة، وتساعدنا على فهم أهمية الأعمدة بالأرقام قبل بناء الموديل.

 Task 5 – Life Probe Sensors
 اختيار أفضل ثلاثة حساسات للمسبار الفضائي

الهدف هو تحليل بيانات 300 مسبار فضائي، ومعرفة الحساسات الأكثر ارتباطًا باكتشاف علامات الحياة، ثم اختيار أفضل ثلاثة حساسات للمسبار القادم بناءً على الأدلة الرقمية.

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(2026)
n = 300

life = rng.integers(0, 2, n)

probes = pd.DataFrame({
    "probe_id": [f"P{i:03d}" for i in range(n)],
    "mission_code": "AX-7",
    "methane_ppm": np.round(5 + life * 12 + rng.normal(0, 3, n), 1),
    "temperature_c": np.round(-40 + life * 25 + rng.normal(0, 15, n), 1),
    "radiation": np.round(rng.normal(100, 10, n), 1),
    "humidity": np.round(20 + life * 6 + rng.normal(0, 8, n), 1),
    "life_detected": life,
})

probes["temperature_f"] = probes["temperature_c"] * 9 / 5 + 32

print("Data created successfully!")
print("Shape:", probes.shape)
display(probes.head())
```

# 1. تطبيق قواعد Filter

في هذه الخطوة أفحص الأعمدة لتحديد العمود التعريفي والعمود الثابت. سأستبعد الأعمدة التي لا تقدم معلومات مفيدة للتحليل، مع الاحتفاظ بقياسات الحرارة مؤقتًا لمقارنة التكرار بينها.

```python
print("عدد القيم الفريدة في كل عمود:")
print(probes.nunique())

probes = probes.drop(columns=["probe_id", "mission_code"])

print("\nShape after filtering:")
print(probes.shape)

display(probes.head())
```

 2. حساب الارتباط مع Target

أحسب معامل الارتباط بين كل حساس وعمود `life_detected` لمعرفة أي الحساسات يرتبط أكثر باكتشاف علامات الحياة.

```python
sensor_cols = [
    "methane_ppm",
    "temperature_c",
    "radiation",
    "humidity",
    "temperature_f"
]

correlations = (
    probes[sensor_cols + ["life_detected"]]
    .corr()["life_detected"]
    .drop("life_detected")
    .sort_values(ascending=False)
)

print("Correlation with life_detected:")
display(correlations)
```

. اكتشاف الحساسات المتكررة

أتحقق من العلاقة بين درجتي الحرارة بالسيلسيوس والفهرنهايت لمعرفة هل يقيسان الشيء نفسه، وهل نحتاج إلى الاحتفاظ بكليهما.

```python
print("Correlation between temperature_c and temperature_f:")
print(probes["temperature_c"].corr(probes["temperature_f"]))
```

 القرار بشأن الحساسات المتكررة

وجدت أن الارتباط بين temperature_c و temperature_f يساوي تقريبًا 1، وهذا يعني أنهما يقيسان الشيء نفسه. لذلك سأحذف temperature_f وأحتفظ بـ temperature_c لتجنب تكرار المعلومات.

```python
probes = probes.drop(columns=["temperature_f"])

sensor_cols = [
    "methane_ppm",
    "temperature_c",
    "radiation",
    "humidity"
]

print("Remaining columns:")
print(probes.columns.tolist())
```

 مقارنة متوسط قراءات الحساسات

أقارن متوسط قراءة كل حساس عندما تكون الحياة مكتشفة وعندما لا تكون مكتشفة، لمعرفة أي الحساسات يظهر فرقًا واضحًا بين الحالتين.

```python
print("Average sensor readings by life_detected:")

display(
    probes.groupby("life_detected")[sensor_cols]
    .mean()
    .T.rename(columns={
        0: "No life",
        1: "Life detected"
    })
)
```

 4. القرار النهائي – اختيار أفضل ثلاثة حساسات

بعد تحليل الارتباط ومقارنة المتوسطات، اخترت الحساسات التالية للمسبار القادم:

1. methane_ppm: الأولوية الأولى؛ لأن معامل ارتباطه مع اكتشاف الحياة هو 0.882، وارتفع متوسطه من 5.38 إلى 17.04 عند اكتشاف الحياة.

2. temperature_c: الأولوية الثانية؛ لأن معامل ارتباطه 0.643، وارتفع متوسطه من -40.31 إلى -14.78 عند اكتشاف الحياة.

3. humidity الأولوية الثالثة؛ لأن معامل ارتباطه 0.397، وارتفع متوسطه من 19.14 إلى 25.63 عند اكتشاف الحياة.

الحساسات والأعمدة المستبعدة
- radiation: استبعدته لأن ارتباطه ضعيف (0.110)، والفرق بين متوسط قراءاته في الحالتين بسيط.
- temperature_f: حذفته لأنه يقيس درجة الحرارة نفسها التي يقيسها temperature_c، والارتباط بينهما يساوي تقريبًا 1.
- probe_id: عمود تعريفي لا يقدم قياسًا مفيدًا.
- mission_code: عمود ثابت لا يحتوي على معلومات متغيرة.
- life_detected: هو العمود المستهدف (Target) وليس حساسًا.

بناءً على هذه النتائج، أوصي باستخدام methane_ppm وtemperature_c وhumidity في المسبار القادم.

 5. Thinking – طريقة التحقق من اختيار الحساسات

سأستخدم طريقة Wrapper وتحديدًا RFECV (Recursive Feature Elimination with Cross-Validation).

اخترت هذه الطريقة لأنها تختبر مجموعات مختلفة من الحساسات، وتقارن أداء النموذج باستخدام التحقق المتقاطع (Cross-Validation)، ثم تساعد في تحديد أفضل الحساسات.

هذه الطريقة مناسبة لأن عدد الحساسات المتاحة قليل، ويمكن استخدامها للتحقق من أن الحساسات الثلاثة المختارة تقدم أفضل أداء للنموذج.

 6. ترتيب الحساسات حسب الأهمية

أرتب الحساسات حسب قوة ارتباطها باكتشاف الحياة، من الأعلى إلى الأقل، لتحديد أولويات استخدامها في المسبار القادم.

```python
sensor_correlations = (
    probes[sensor_cols + ["life_detected"]]
    .corr()["life_detected"]
    .drop("life_detected")
    .sort_values(ascending=False)
)

print("Sensors ranked by correlation:")
display(sensor_correlations)
```
