تحليل مدة القرض Loan_Amount_Term

لاحظت أن نسبة الموافقة على القروض للصفوف التي تفتقد مدة القرض بلغت 57.14%. كما لاحظت اختلافًا واضحًا في نسبة الموافقة حسب Credit_History؛ إذ بلغت 7.87% عندما كانت القيمة 0، و79.58% عندما كانت القيمة 1.

قرار معالجة القيم الناقصة في Loan_Amount_Term

لاحظت أن مدة القرض الأكثر تكرارًا هي 360 شهرًا، أي 30 سنة، وقد تكررت 512 مرة. لذلك سأستخدم هذه القيمة لتعويض القيم الناقصة؛ لأنها الأكثر شيوعًا في البيانات.

```python
for col in ["ApplicantIncome", "CoapplicantIncome", "LoanAmount"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print("\nColumn:", col)
    print("Lower limit:", lower)
    print("Upper limit:", upper)
    print("Outliers count:", len(outliers))
```

# Task 5  سجلات متحف الديناصورات

```python
import pandas as pd

fossils = {
    "fossil_id": ["F01", "F02", "F03", "F04", "F05", "F02", "F06", "F07", "F08", "F09"],
    "species": ["T-Rex", "Triceratops", "t-rex", "Velociraptor", "Stegosaurus", "Triceratops",
                "Brachiosaurus", "T-Rex", "Velociraptor", "Stegosaurus"],
    "age_million_years": [68, 67, 66, 75, -150, 67, 150, 68, None, 155],
    "length": ["12 m", "9 m", "40 ft", "2 m", "9 m", "9 m", "26 m", "12.3 m", "6.6 ft", "9 m"],
    "found_in": ["USA", "USA", "Canada", "Mongolia", "USA", "USA", "Tanzania", "USA", "Mongolia",
                 "Portugal"],
}

fossils_df = pd.DataFrame(fossils)

print(fossils_df)
```

```python
fossils_df = fossils_df.drop_duplicates()

fossils_df["species"] = fossils_df["species"].str.title()

print(fossils_df)
```

```python
fossils_df.loc[fossils_df["age_million_years"] < 0, "age_million_years"] = 150

print(fossils_df[["species", "age_million_years"]])
```

# سبب العمر السالب

السبب الأرجح هو خطأ أثناء إدخال البيانات، مثل وضع إشارة السالب بدل الموجب. لذلك صححت العمر من -150 إلى 150 مليون سنة.

```python
fossils_df["age_million_years"] = fossils_df["age_million_years"].fillna(
    fossils_df.groupby("species")["age_million_years"].transform("median")
)

print(fossils_df[["species", "age_million_years"]])
```

### لماذا استخدام وسيط النوع نفسه أدق؟

استخدام وسيط عمر النوع نفسه أدق؛ لأن أعمار الديناصورات تختلف من نوع إلى آخر. لذلك تعويض العمر بناءً على النوع نفسه يعطي قيمة أقرب لبياناته، بدل استخدام وسيط جميع الحفريات.

```python
split_length = fossils_df["length"].str.split(" ", expand=True)

fossils_df["length_value"] = split_length[0].astype(float)
fossils_df["length_unit"] = split_length[1]

fossils_df["length_m"] = fossils_df["length_value"]

fossils_df.loc[
    fossils_df["length_unit"] == "ft", "length_m"
] *= 0.3048

print(fossils_df[["length", "length_value", "length_unit", "length_m"]])
```

```python
print("Average length for each species:")
print(fossils_df.groupby("species")["length_m"].mean().sort_values())

print("\nOldest species:")
oldest = fossils_df.loc[fossils_df["age_million_years"].idxmax()]
print(oldest["species"])
print(oldest["age_million_years"])

print("\nCountry with the most fossils:")
print(fossils_df["found_in"].value_counts().head(1))
```
