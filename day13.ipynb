مثال من الحياة: إذا كان متوسط درجاتي في الاختبارات 80، فهذا لا يعني أن جميع درجاتي متقاربة؛ فقد أحصل على 100 في اختبار و60 في اختبار آخر. هنا المتوسط أعطاني رقمًا واحدًا، لكنه لم يوضح الاختلاف بين الدرجات.



 2. اختيار الرسمة المناسبة

| السؤال | نوع الرسمة | لماذا اخترتها؟ |
|---|---|---|
| كيف تغيرت مبيعات المقهى خلال 12 شهرًا؟ | Line Chart (رسم خطي) | لأنها توضح تغير المبيعات والارتفاع والانخفاض خلال الأشهر. |
| أي فرع باع أكثر هذا الشهر من 5 فروع؟ | Bar Chart (رسم أعمدة) | لأنها تساعد على مقارنة المبيعات بين الفروع وتحديد الفرع الأعلى مبيعًا. |
| كيف تتوزع أعمار 500 عميل؟ | Histogram (مدرج تكراري) | لأنه يوضح توزيع الأعمار وعدد العملاء في كل فئة عمرية. |
| هل يزيد الإيجار كلما زادت مساحة الشقة؟ | Scatter Plot (مخطط مبعثر) | لأنه يوضح العلاقة بين مساحة الشقة وقيمة الإيجار. |
| هل في رواتب الموظفين قيم شاذة؟ | Box Plot (مخطط الصندوق) | لأنه يساعد على اكتشاف القيم الشاذة والرواتب البعيدة عن بقية الرواتب. |


 3. أجزاء الرسمة

يوجد فرق بين Figure وAxes:

- Figure: هو الشكل الكامل الذي يحتوي على الرسمة أو عدة رسومات.
- Axes: هي المنطقة التي يتم فيها عرض البيانات ورسم المحاور.

وظيفة أجزاء الرسمة:

- Title: يوضح موضوع الرسمة وماذا تعرض.
- X-axis: يوضح المعلومات الموجودة على المحور الأفقي.
- Y-axis: يوضح المعلومات الموجودة على المحور الرأسي.
- Legend: يوضح معنى الألوان والخطوط والرموز المستخدمة في الرسمة.

إذا أرسلت للمدير رسمة بدون هذه العناصر، فقد يصعب عليه فهم البيانات ومعرفة ماذا تمثل الأرقام والخطوط، وقد يؤدي ذلك إلى تفسير خاطئ للنتائج. لذلك يجب أن تكون الرسمة واضحة وتحتوي على جميع العناصر المهمة.


# 4. الرسمة المضللة

الخدعة هي أن محور Y لا يبدأ من الصفر، بل يبدأ من قيمة قريبة من 94%، وهذا يجعل الفرق بين 94% و96% يبدو أكبر بكثير من حجمه الحقيقي، وكأن رضا العملاء تضاعف.

لكن في الحقيقة الفرق بين السنتين هو نقطتان مئويتان فقط، وليس ضعفًا.

لإصلاح الرسمة، يجب أن يبدأ المحور الرأسي من الصفر، مع توضيح القيم والعنوان وتسميات المحاور؛ حتى تظهر المقارنة بشكل عادل ولا تعطي القارئ انطباعًا خاطئًا.

 Task 2: تطبيق الرسوم البيانية على بيانات المقهى

```python
import matplotlib.pyplot as plt

days = ["Sat", "Sun", "Mon", "Tue", "Wed", "Thu", "Fri"]
latte = [42, 38, 35, 40, 44, 58, 61]
espresso = [30, 28, 33, 31, 29, 35, 22]

drinks = ["Latte", "Espresso", "Mocha", "Tea"]

ratings = [5, 4, 4, 5, 3, 5, 4, 2, 5, 4, 4, 5, 3, 4, 5, 5, 1, 4, 4, 5,
           3, 4, 5, 4, 2, 5, 4, 4, 3, 5, 4, 5, 5, 4, 3, 4, 5, 4, 4, 5]

temp = [18, 22, 25, 27, 30, 33, 35, 38, 40, 41]
iced = [5, 9, 12, 15, 20, 26, 30, 35, 39, 38]
```

 الرسم الأول: Line Chart

الرسم الخطي يستخدم لمتابعة التغير في البيانات مع مرور الوقت. سنستخدمه لمقارنة مبيعات Latte وEspresso خلال أيام الأسبوع.

- `plt.plot()` يرسم خطًا يمثل البيانات.
- `marker` يضيف علامة عند كل نقطة بيانات.
- `label` يحدد اسم كل خط.
- `plt.legend()` يعرض أسماء الخطوط في الرسم.
- `plt.title()` يضيف عنوان الرسم.
- `plt.xlabel()` يضيف اسم المحور الأفقي.
- `plt.ylabel()` يضيف اسم المحور الرأسي.

```python
plt.figure(figsize=(10, 5))

plt.plot(days, latte, marker='o', label='Latte')
plt.plot(days, espresso, marker='s', label='Espresso')

plt.title('Latte and Espresso Sales During the Week')
plt.xlabel('Days')
plt.ylabel('Number of Sales')

plt.legend()
plt.grid(True)

plt.show()
```

 تحليل الرسم البياني

نلاحظ أن مبيعات Latte وصلت إلى أعلى مستوى يوم الجمعة بمقدار 61 عملية بيع، كما ارتفعت بشكل واضح في يومي الخميس والجمعة.

أما مبيعات Espresso فقد انخفضت يوم الجمعة إلى 22 عملية بيع بعد أن كانت 35 يوم الخميس.

نستنتج أن Latte كان المشروب الأكثر مبيعًا خلال الأسبوع، بينما انخفض الطلب على Espresso في نهاية الأسبوع.

 الرسم الثاني: Bar Chart

الرسم بالأعمدة يستخدم لمقارنة القيم بين فئات مختلفة. سنستخدمه لمقارنة إجمالي مبيعات المشروبات الأربعة خلال الأسبوع.

- `sum()` يحسب مجموع المبيعات.
- `plt.bar()` يرسم الأعمدة.
- `plt.text()` يكتب قيمة المبيعات فوق كل عمود.
- `plt.title()` يضيف عنوان الرسم.
- `plt.xlabel()` يضيف اسم المحور الأفقي.
- `plt.ylabel()` يضيف اسم المحور الرأسي.

```python
latte_total = sum(latte)
espresso_total = sum(espresso)

sales = [latte_total, espresso_total, 145, 96]

print("Latte:", latte_total)
print("Espresso:", espresso_total)
print("Mocha:", 145)
print("Tea:", 96)
```

```python
plt.figure(figsize=(8, 5))

bars = plt.bar(drinks, sales)

plt.title('Total Weekly Sales by Drink')
plt.xlabel('Drink')
plt.ylabel('Total Sales')

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        str(height),
        ha='center',
        va='bottom'
    )

plt.show()
```

نلاحظ أن مشروب اللاتيه حقق أعلى إجمالي مبيعات خلال الأسبوع بمجموع 318 عملية بيع، يليه الإسبريسو بمجموع 208 عمليات بيع، ثم الموكا بمجموع 145 عملية بيع.

أما الشاي فحقق أقل إجمالي مبيعات بمجموع 96 عملية بيع.

نستنتج أن اللاتيه هو المشروب الأكثر مبيعًا خلال الأسبوع، بينما الشاي هو الأقل مبيعًا.

 الرسم الثالث: Histogram

المدرج التكراري يستخدم لمعرفة توزيع البيانات وعدد مرات تكرار كل قيمة.

- `plt.hist()` يرسم توزيع التقييمات.
- `bins` يحدد عدد الفئات في الرسم.
- `plt.title()` يضيف عنوان الرسم.
- `plt.xlabel()` يضيف اسم المحور الأفقي.
- `plt.ylabel()` يضيف اسم المحور الرأسي.

**التوقع قبل تشغيل الرسم:**

أتوقع أن التقييم 5 هو الأكثر تكرارًا؛ لأن معظم التقييمات في البيانات تبدو مرتفعة.

```python
plt.figure(figsize=(8, 5))

plt.hist(ratings, bins=[0.5, 1.5, 2.5, 3.5, 4.5, 5.5], edgecolor='black')

plt.title('Distribution of Customer Ratings')
plt.xlabel('Rating')
plt.ylabel('Frequency')

plt.xticks([1, 2, 3, 4, 5])
plt.show()
```

: Scatter

المخطط المبعثر يستخدم لمعرفة العلاقة بين متغيرين، ويساعدنا على معرفة ما إذا كان هناك ارتباط بينهما.

- `plt.scatter()` يرسم النقاط التي توضح العلاقة بين المتغيرين.
- `plt.title()` يضيف عنوان الرسم.
- `plt.xlabel()` يضيف اسم المحور الأفقي.
- `plt.ylabel()` يضيف اسم المحور الرأسي.
- `plt.grid()` يضيف شبكة لتسهيل قراءة الرسم.

```python
plt.figure(figsize=(8, 5))

plt.scatter(temp, iced, marker='o')

plt.title('Temperature vs Iced Coffee Sales')
plt.xlabel('Temperature')
plt.ylabel('Iced Coffee Sales')

plt.grid(True)
plt.show()
```

 تحليل الرسم البياني

نلاحظ وجود علاقة طردية بين درجة الحرارة ومبيعات القهوة المثلجة؛ فكلما ارتفعت درجة الحرارة زادت المبيعات بشكل عام.

وصلت المبيعات إلى أعلى مستوى عند درجة حرارة 40، ثم انخفضت قليلًا عند درجة حرارة 41.

نستنتج أن ارتفاع درجة الحرارة يزيد من طلب العملاء على القهوة المثلجة.

  : Subplots

تستخدم Subplots لعرض أكثر من رسم بياني في شكل واحد، مما يسهل مقارنة البيانات.

- `plt.subplots()` ينشئ عدة رسومات في شكل واحد.
- `axs[0, 0]` يحدد مكان الرسم الأول.
- `axs[0, 1]` يحدد مكان الرسم الثاني.
- `axs[1, 0]` يحدد مكان الرسم الثالث.
- `axs[1, 1]` يحدد مكان الرسم الرابع.
- `fig.suptitle()` يضيف عنوانًا رئيسيًا للوحة.
- `plt.tight_layout()` ينظم المسافات بين الرسومات.
- `plt.savefig()` يحفظ الصورة باسم محدد.

```python
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# الرسم الأول: Line Chart
axs[0, 0].plot(days, latte, marker='o', label='Latte')
axs[0, 0].plot(days, espresso, marker='s', label='Espresso')
axs[0, 0].set_title('Latte and Espresso Sales')
axs[0, 0].set_xlabel('Days')
axs[0, 0].set_ylabel('Number of Sales')
axs[0, 0].legend()
axs[0, 0].grid(True)

# الرسم الثاني: Bar Chart
bars = axs[0, 1].bar(drinks, sales)
axs[0, 1].set_title('Total Weekly Sales by Drink')
axs[0, 1].set_xlabel('Drink')
axs[0, 1].set_ylabel('Total Sales')

for bar in bars:
    height = bar.get_height()
    axs[0, 1].text(
        bar.get_x() + bar.get_width()/2,
        height,
        str(height),
        ha='center',
        va='bottom'
    )

# الرسم الثالث: Histogram
axs[1, 0].hist(
    ratings,
    bins=[0.5, 1.5, 2.5, 3.5, 4.5, 5.5],
    edgecolor='black'
)
axs[1, 0].set_title('Customer Ratings Distribution')
axs[1, 0].set_xlabel('Rating')
axs[1, 0].set_ylabel('Frequency')
axs[1, 0].set_xticks([1, 2, 3, 4, 5])

# الرسم الرابع: Scatter
axs[1, 1].scatter(temp, iced, marker='o')
axs[1, 1].set_title('Temperature vs Iced Coffee Sales')
axs[1, 1].set_xlabel('Temperature')
axs[1, 1].set_ylabel('Iced Coffee Sales')
axs[1, 1].grid(True)

# العنوان الرئيسي
fig.suptitle('Cafe Sales Dashboard', fontsize=16)

plt.tight_layout(rect=[0, 0, 1, 0.96])

# حفظ الصورة
plt.savefig('cafe_dashboard.png', dpi=300)

plt.show()
```

```python
import zipfile

with zipfile.ZipFile("archive (6).zip", "r") as z:
    z.extract("insurance.csv")

import pandas as pd

insurance_df = pd.read_csv("insurance.csv")

print("Shape:", insurance_df.shape)
print("Columns:", insurance_df.columns.tolist())
```

```python
print("Duplicate rows:", insurance_df.duplicated().sum())

insurance_df = insurance_df.drop_duplicates()

print("Shape after removing duplicates:", insurance_df.shape)
```

 الرسم الأول: Histogram لتوزيع تكاليف التأمين

يُستخدم الرسم المدرج التكراري (Histogram) لفهم توزيع تكاليف التأمين الطبي (`charges`)، ومعرفة القيم الأكثر تكرارًا وشكل انتشار التكاليف.

سنضيف خطًا للمتوسط الحسابي (Mean) وخطًا للوسيط (Median) باستخدام `plt.axvline()`، حتى نقارن بينهما ونفهم تأثير القيم المرتفعة على التوزيع.

```python
import matplotlib.pyplot as plt

mean_charges = insurance_df["charges"].mean()
median_charges = insurance_df["charges"].median()

plt.figure(figsize=(10, 6))

plt.hist(insurance_df["charges"], bins=30, edgecolor="black", alpha=0.7)

plt.axvline(mean_charges, color="red", linestyle="--",
            label=f"Mean = {mean_charges:.2f}")

plt.axvline(median_charges, color="green", linestyle="-",
            label=f"Median = {median_charges:.2f}")

plt.title("Distribution of Medical Insurance Charges")
plt.xlabel("Insurance Charges")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

 ماذا يخبرنا شكل التوزيع عن الفرق بين المتوسط والوسيط؟

يظهر توزيع تكاليف التأمين انحرافًا نحو اليمين بسبب وجود بعض القيم المرتفعة. بلغ المتوسط الحسابي 13,279.12، بينما بلغ الوسيط 9,386.16. وهذا يعني أن التكاليف المرتفعة أثرت على المتوسط ورفعته، بينما الوسيط أقل تأثرًا بها.

 الرسم الثاني: Scatter Plot للعمر وتكاليف التأمين

يُستخدم الرسم المبعثر (Scatter Plot) لدراسة العلاقة بين عمر الشخص وتكاليف التأمين الطبي.

سنستخدم لونًا مختلفًا للمدخنين وغير المدخنين، حتى نلاحظ الفرق بين مجموعتي البيانات ونفهم كيف يرتبط التدخين بتكاليف التأمين.

```python
plt.figure(figsize=(10, 6))

smokers = insurance_df[insurance_df["smoker"] == "yes"]
non_smokers = insurance_df[insurance_df["smoker"] == "no"]

plt.scatter(non_smokers["age"], non_smokers["charges"],
            color="blue", alpha=0.6, label="Non-Smokers")

plt.scatter(smokers["age"], smokers["charges"],
            color="red", alpha=0.6, label="Smokers")

plt.title("Age vs Medical Insurance Charges")
plt.xlabel("Age")
plt.ylabel("Insurance Charges")
plt.legend()
plt.show()
```

 ماذا نلاحظ من الرسم؟

نلاحظ وجود نطاقين (Bands) في الرسم، أحدهما للمدخنين والآخر لغير المدخنين.

يظهر أن المدخنين يتركزون في النطاق العلوي بتكاليف تأمين أعلى، بينما يتركز غير المدخنين في النطاق السفلي بتكاليف أقل.

بلغ متوسط تكاليف التأمين للمدخنين 32,050.23، مقارنةً بـ 8,440.66 لغير المدخنين. وهذا يوضح وجود فرق كبير بين المجموعتين.

  مقارنة متوسط تكاليف التأمين

سنستخدم Bar Chart لمقارنة متوسط تكاليف التأمين الطبي حسب حالة التدخين، ومقارنة متوسط التكاليف بين المناطق المختلفة.

سنعرض الرسمين بجانب بعض باستخدام `plt.subplots()` لتسهيل المقارنة بين المجموعات.

```python
smoker_charges = insurance_df.groupby("smoker")["charges"].mean()
region_charges = insurance_df.groupby("region")["charges"].mean()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

smoker_charges.plot(kind="bar", ax=axes[0], color=["skyblue", "salmon"])
axes[0].set_title("Average Charges by Smoking Status")
axes[0].set_xlabel("Smoking Status")
axes[0].set_ylabel("Average Charges")

region_charges.plot(kind="bar", ax=axes[1], color="teal")
axes[1].set_title("Average Charges by Region")
axes[1].set_xlabel("Region")
axes[1].set_ylabel("Average Charges")

plt.tight_layout()
plt.show()
```

 ملاحظات على مقارنة متوسط تكاليف التأمين

يوضح الرسم أن متوسط تكاليف التأمين للمدخنين أعلى بكثير من غير المدخنين؛ حيث بلغ 32,050.23 للمدخنين، مقابل 8,440.66 لغير المدخنين.

كما تختلف متوسطات التكاليف بين المناطق، وكانت منطقة Southeast الأعلى بمتوسط 14,735.41، بينما كانت منطقة Southwest الأقل بمتوسط 12,346.94.

 : الرسم المضلل (Misleading Graph)

سنرسم متوسط تكاليف التأمين حسب المنطقة مرتين باستخدام البيانات نفسها، لكن مع تغيير بداية المحور الرأسي (Y-axis).

في الرسم الأول سيبدأ المحور من 12,000، وفي الرسم الثاني سيبدأ من صفر. الهدف هو ملاحظة كيف يمكن أن يؤدي تغيير مقياس المحور إلى إعطاء انطباع مختلف عن الفروقات الحقيقية بين المناطق.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# الرسم الأول: المحور يبدأ من 12000
region_charges.plot(kind="bar", ax=axes[0], color="skyblue")
axes[0].set_ylim(bottom=12000)
axes[0].set_title("Misleading Graph: Y-axis Starts at 12000")
axes[0].set_xlabel("Region")
axes[0].set_ylabel("Average Charges")

# الرسم الثاني: المحور يبدأ من صفر
region_charges.plot(kind="bar", ax=axes[1], color="teal")
axes[1].set_ylim(bottom=0)
axes[1].set_title("Accurate Graph: Y-axis Starts at Zero")
axes[1].set_xlabel("Region")
axes[1].set_ylabel("Average Charges")

plt.tight_layout()
plt.show()
```

# Task 4 محطة أرصاد المريخ
مقدمة المهمة

في هذه المهمة سنحلل قراءات محطة الأرصاد على المريخ لمدة 60 يومًا مريخيًا (Sol)، بهدف اختيار أفضل نافذة لإطلاق مركبة فضائية.

سندرس درجات الحرارة وسرعة الرياح وأيام العواصف الترابية باستخدام الرسوم البيانية، ثم نبحث عن أفضل خمسة أيام متتالية تكون خالية من العواصف وأكثر دفئًا.

# سنستخدم الرسوم البيانية والتحليل البرمجي لدعم قرار اختيار موعد الإطلاق.

```python
import numpy as np

rng = np.random.default_rng(2026)

sol = np.arange(1, 61)

temp = np.round(-60 + 15 * np.sin(sol / 9) + rng.normal(0, 4, 60), 1)

storm = np.zeros(60, dtype=int)
storm[[11, 12, 13, 32, 33, 46, 47, 48, 49]] = 1

temp[storm == 1] -= 12

wind = np.round(20 + storm * 35 + rng.normal(0, 5, 60), 1)
```

 الرسم الأول: درجات الحرارة والرياح على المريخ

سنرسم درجات الحرارة خلال الأيام المريخية باستخدام Line Chart، ونضيف نقاطًا حمراء لتحديد أيام العواصف الترابية باستخدام Scatter Plot.

سنضيف أيضًا خطًا أفقيًا يمثل متوسط درجات الحرارة، وسهمًا يشير إلى أبرد يوم مع رقم اليوم.

وفي الرسم الثاني سنعرض سرعة الرياح باستخدام أعمدة حمراء في أيام العواصف، حتى نتمكن من مقارنة الحرارة والرياح على نفس المحور الزمني.

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 1, figsize=(14, 9), sharex=True)

# الرسم الأول: درجات الحرارة
axes[0].plot(sol, temp, color="blue", marker="o", label="Temperature")

# نقاط حمراء لأيام العواصف
axes[0].scatter(
    sol[storm == 1],
    temp[storm == 1],
    color="red",
    label="Dust Storm"
)

# خط متوسط الحرارة
mean_temp = np.mean(temp)
axes[0].axhline(mean_temp, color="green", linestyle="--",
                label=f"Mean Temperature = {mean_temp:.2f}")

# تحديد أبرد يوم
coldest_index = np.argmin(temp)
axes[0].annotate(
    f"Coldest Day: Sol {sol[coldest_index]}",
    xy=(sol[coldest_index], temp[coldest_index]),
    xytext=(sol[coldest_index] + 5, temp[coldest_index] + 10),
    arrowprops=dict(arrowstyle="->", color="black")
)

axes[0].set_title("Temperature on Mars")
axes[0].set_ylabel("Temperature (C)")
axes[0].legend()

# الرسم الثاني: سرعة الرياح
wind_colors = np.where(storm == 1, "red", "skyblue")

axes[1].bar(sol, wind, color=wind_colors)
axes[1].set_title("Wind Speed on Mars")
axes[1].set_xlabel("Martian Sol")
axes[1].set_ylabel("Wind Speed")
axes[1].legend(["Wind Speed"])

plt.tight_layout()
plt.show()
```

 اختيار أفضل نافذة لإطلاق المركبة

من خلال الرسم البياني، سأبحث عن خمسة أيام مريخية متتالية تكون خالية من العواصف الترابية، وتتميز بأعلى درجات حرارة ممكنة مقارنة ببقية الأيام.

بعد ذلك سأستخدم حلقة `for` وعمليات `slicing` لفحص جميع النوافذ الممكنة واختيار الأفضل.

من خلال الرسم البياني، تبدو الأيام من Sol 15 إلى Sol 19 مناسبة للإطلاق؛ لأنها خالية من العواصف وتتميز بدرجات حرارة أعلى من كثير من الأيام الأخرى.

```python
best_avg = -float("inf")
best_start = None

for i in range(len(sol) - 4):

    # اختيار خمسة أيام متتالية باستخدام slicing
    temp_window = temp[i:i+5]
    storm_window = storm[i:i+5]
    sol_window = sol[i:i+5]

    # التأكد من عدم وجود عواصف
    if np.all(storm_window == 0):

        # حساب متوسط الحرارة
        avg_temp = np.mean(temp_window)

        # اختيار النافذة الأدفأ
        if avg_temp > best_avg:
            best_avg = avg_temp
            best_start = i

# عرض أفضل نافذة
print("Best Launch Window:")
print("From Sol:", sol[best_start])
print("To Sol:", sol[best_start + 4])
print("Average Temperature:", round(best_avg, 2))
print("Temperatures:", temp[best_start:best_start+5])
```

بعد تحليل درجات الحرارة وأيام العواصف الترابية، اخترت الأيام من Sol 15 إلى Sol 19 كأفضل نافذة لإطلاق المركبة.

بلغ متوسط درجة الحرارة خلال هذه الأيام -46.3 درجة مئوية، وكانت الأيام الخمسة خالية من العواصف الترابية.

اخترت هذه النافذة لأنها تجمع بين عدم وجود العواصف وارتفاع درجات الحرارة مقارنة بالنوافذ الأخرى التي تم فحصها.
